## Project Goal:
This project seeks to determine if  brain pH is correlated with tau and amyloid-beta concentrations, and whether the mean values of the concentrations of these proteins differ between patients with and without atherosclerosis.

## Disease Background:

* Prevalence & incidence:

It’s estimated that approximately 7.4 million Americans have Alzheimer’s disease, about 11% of all people aged 65 and older. The prevalence by age group is 5.2% of all people aged 65-74, 13.8% of 75-84 year olds, all the way up to 35.8% of people over the age of 85. It’s unlikely, but not impossible, for Alzheimer’s symptoms to develop under the age of 65 because of the significant impact between again and the disease. Based on the growth rate of the prevalence, it’s also estimated that the number will almost double every 20 years.<br>

* Economic burden:

According to a study published in PubMed, the annual per capita cost of Alzheimer’s disease ranges from $468.28 to $171,283.80, based on severity. The vast majority of this figure comes from the cost of caregiving, because people with more severe Alzheimer’s are often incapable of caring for themselves. The majority of these costs are paid by Medicare/Medicaid, with only a small amount covered by private insurance, and the remaining out of pocket. Caring for a relative with Alzheimer's is a significant economic burden, and can often require around-the-clock care, causing someone to either have to hire an expensive caregiver or sacrifice work hours or even leave their jobs to be able to take care of their loved ones.<br>

* Risk factors (genetic, lifestyle):

Because we don’t know exactly how Alzheimer’s starts, we can’t say for certain what causes it, but there are a number of risk factors. Some of the most influential ones are age, genetics, lifestyle, and overall health. Age is the most significant risk factor, with 11% of people 65% affected. Genetics can also impact a person's likelihood of developing Alzheimer's, as variants of many genes have been found that increase or decrease the risk. In addition, exercise, diet, and cognitive engagement can reduce risk of Alzheimer’s by improving overall brain health. <br>
https://www.alz.org/getmedia/ef8f48f9-ad36-48ea-87f9-b74034635c1e/alzheimers-facts-and-figures.pdf
* Societal determinants:

Lower socioeconomic status can increase risk of Alzheimer’s, mainly because of the overall health impacts. As discussed above, healthier lifestyles can prevent Alzheimer’s, and lower socioeconomic status means less access to exercise, healthy food, medecine, and more. In addition, those with more years of formal education are at lower risk of developing Alzheimer's because they have healthier brains. Another, somewhat lesser known, societal determinant is living environment, specifically exposure to fine particulate matter air pollution. Recent studies have reported higher long-term exposure is correlated to worse cognitive decline and severity of Alzheimer's.<br>
https://www.alz.org/getmedia/ef8f48f9-ad36-48ea-87f9-b74034635c1e/alzheimers-facts-and-figures.pdf
* Symptoms:

In Alzheimer’s disease, neurons gradually break down because of the build up of plaque. This causes cognitive impairments such as memory problems, emotional dysfunction, communication problems, and others. In early stages, Alzheimer's generally presents with minor cognitive decline and subtle changes in mood or behavior. By later stages of the disease symptoms can even include difficulty walking, speaking, and swallowing because of the damage to the cerebral cortex.<br>
https://www.alz.org/getmedia/ef8f48f9-ad36-48ea-87f9-b74034635c1e/alzheimers-facts-and-figures.pdf
* Diagnosis:

To diagnose AD, commonly patients take memory tests, such as the Montreal Cognitive Assessment (MoCA test) and the Mini-Mental State Exam (MMSE) listed on the Alzheimer’s Metadata sheet used for this project. 
In the more unusual case that the patient is young and suspected with early-onset AD, a gene test may help assess the likelihood of AD, with stronger chances correlated to the APOE-ε4 alleles.
To further diagnose the patient, given they have shown signs of cognitive impairment, the patient can undergo brain imaging. An atrophied hippocampus, shown via an MRI scan, would indicate AD, and CT scans (in addition to MRI scans) can rule out other possible causes of cognitive impairment such as a brain tumor. Another common imaging tool is the PET scan, which can show the concentration of amyloid beta plaques and tau tangles in the brain (via an injected dye that selectively binds to those proteins and relays a radioactive signal to the machine to record). 
Alternatively, the doctor could assess the content of AD-associated proteins in the patients’ brains via spinal taps that allow the doctor to sample the contents of the patients’ brain fluid (aka cerebrospinal fluid). 
Recently, the FDA approved a blood test called Lumipulse that is cited to be 90% effective at picking up on the abnormally high concentrations of amyloid beta and tau proteins in patients with AD.

* Standard of care treatments (& reimbursement):

First, doctors may prescribe cholinesterase inhibitors (donepezil, galantamine, or rivastigmine transdermal patch), which help block the enzyme from breaking down acetylcholine—an already lacking neurotransmitter in AD patients. Another common drug is Memantine; this one works by inhibiting neuronal NMDA receptors—which are overstimulated by glutamate in AD-affected brains. A third class of drugs include monoclonal antibodies, Lecanemab and Donanemab, that have been tested to delay cognitive worsening by 27% and 35%, respectively, by binding to the amyloid beta plaques and marking them as targets for immune cells to clear. Unfortunately, none of these drugs can cure AD, and they all have their side-effects. 
Although the cholinesterase inhibitors and Memantine are relatively cheap, the monoclonal antibody drugs can cost tens of thousands of dollars per year! Along with private insurance, medicare covers these drugs in parts B or D, but nonetheless, this costs major taxpayer money and really makes insurance more expensive for everybody. What doesn’t cost as much is exercise, nutrition (Vitamin E, Omega-3 fatty acids, antioxidants, protein, etc), light therapy, and sleep!

* Disease progression & prognosis:

After diagnosis, patients with AD generally only have 3-10 years of life left, with the average expectancy decreasing with age. Even before symptoms and subsequent diagnosis, AD-associated amyloid beta plaques and tau tangles have already been accumulating during this “pre-clinical stage.” As the disease progresses to its mild, moderate, then severe stages, the patient’s memory lapses become more and more frequent and may even develop a different personality—irritable, quiet, or untrusting. Eventually, the patient loses the ability to speak, eat, get dressed, walk, etcetera.

* Continuum of care providers:

Especially for safety reasons, patients with AD need to be frequently looked after. Families, friends, and non-profits called Area Agencies on Aging are common care services; they may help with drug injections, feeding, and general house upkeep. Physical therapists may also be hired to provide mobility sessions, and as the disease progresses to a life-expectancy of six months, hospice workers often come in to help thrice weekly. Alternatively, the family or patient can opt to live in a senior independent living facility, attend adult daycare, or choose to live at a nursing facility—which may even have a specialized Alzheimer's unit that can provide around-the-clock medical services.

https://health.clevelandclinic.org/long-term-care-for-alzheimers-patients 


* Biological mechanisms (anatomy, organ physiology, cell & molecular physiology):

Beginning predominantly in the brain’s bilateral hippocampi regions, it is hypothesized that in patients with AD, enzymes beta-secretase and gamma-secretase catalyze Amyloid Precursor protein into Amyloid Beta at higher rates than in patients without AD. These amyloid-beta proteins congregate and “stick” together outside of the cell, disrupting cell communication. Furthermore, amyloid-beta plaques lead to the excessive hyperphosphorylation of a protein called tau—a key part of the cell’s microtubule system that helps it transport stuff like nutrients, neurotransmitters, and other proteins. The end result is apoptosis, or cell death, and on a broader scale, atrophy of the hippocampi that is responsible for memory.


* Clinical Trials/next-gen therapies:

One next-gen therapy to get excited about includes a non-invasive magnetic stimulation—customized to each patients’ brain network. According to the company website, “Sinaptica’s personalized rTMS-EEG neuromodulation was shown to significantly preserve cognition and function and reduced behavioral disturbances after 12 months, compared to placebo.” The data shows this novel treatment slows cognitive decline by 44%!


Pertaining to drugs, trontinemab’s phase 3 trial in 2028 will be important to watch out for, as prior phase 2 showed a clearing to normal levels of amyloid beta in 92% of patients. While there are other antibodies on the market for amyloid plaque, trontinemab is novel in that it more efficiently crosses the blood-brain barrier.
Lastly, blarcamesine is a medicine in phase 2 trials that uniquely works by activating a “sigma-1 receptor” which is supposed to signal the brain to dispose of the amyloid beta plaque. What’s cool about this emerging treatment is that it’d be offered in pill form, so the patient wouldn’t have to be periodically injected by a trained family member or nurse.

We compared the measure of atherosclerosis severity (none, mild, moderate, and severe) to the concentrations in pg/ug of Amyloid-Beta 40, Amyloid-Beta 42, total Tau, and phosphorylated Tau in a bar graph. This function iterates through every patient with the value of "None" for their atherosclerosis and adds each protein value to the respective list. There are three more near-identical functions for the other qualitative measures of atherosclerosis.

```python
for patient in Patient.filter(Patient.all_patients, atherosclerosis = "None"):
   abeta40_levels_of_patients_with_no_atherosclerosis.append(patient.abeta40_level)
   abeta42_levels_of_patients_with_no_atherosclerosis.append(patient.abeta42_level)
   ttau_levels_of_patients_with_no_atherosclerosis.append(patient.ttau_level)
   ptau_levels_of_patients_with_no_atherosclerosis.append(patient.ptau_level)
```

For each atherosclerosis severity measurement, we calculated the mean and standard deviation of the values for each of the four types of proteins.

```python
x_abeta40_no_atherosclerosis_bar = (statistics.mean(abeta40_levels_of_patients_with_no_atherosclerosis))
abeta40_no_atherosclerosis_stdev = (statistics.stdev(abeta40_levels_of_patients_with_no_atherosclerosis))
```

We then plotted all the data on the same bar graph, with each of the four protein concentrations graphed for the four measures of atherosclerosis severity.

```python
atherosclerosis_condition_cols = ['None', 'Mild', 'Moderate', 'Severe'] # X-axis labels
x = np.arange(len(atherosclerosis_condition_cols)) # Numbered position for each category
width = 0.2
plt.xticks(x, atherosclerosis_condition_cols) 
plt.xlabel("Atherosclerosis")
plt.ylabel("Concentration (pg/ug)")
plt.title("Differences in Amyloid Beta and Tau by Atherosclerosis Severity")
plt.legend(title="Protein") #creates the key

plt.bar(
    x - 1.5*width, #subtracts x value by 0.3 (0.2 times 1.5); this is so each bar representing abeta 40 is displaced 0.3 units from each of the x locations [0,1,2,3]
    [x_abeta40_no_atherosclerosis_bar, #at x = -0.3
     x_abeta40_mild_atherosclerosis_bar, # at x = 0.7
     x_abeta40_moderate_atherosclerosis_bar, #at x = 1.7
     x_abeta40_severe_atherosclerosis_bar], #at x = 2.7
    width, #width of bar
    label="ABeta40",
    yerr=[[0,0,0,0], # so that standard deviation bar only goes up (assymmetrical)
         [abeta40_no_atherosclerosis_stdev,
         abeta40_mild_atherosclerosis_stdev,
         abeta40_moderate_atherosclerosis_stdev,
         abeta40_severe_atherosclerosis_stdev]],
    capsize=5 #width of the cap on the standard deviation bar
)
```
We ran a one-way ANOVA test on the data to look for significant differences between the groups we compared.

```python
# Function runs ANOVA on the following lists
f_abeta40, p_abeta40 = stats.f_oneway(
   abeta40_levels_of_patients_with_no_atherosclerosis,
   abeta40_levels_of_patients_with_mild_atherosclerosis,
   abeta40_levels_of_patients_with_moderate_atherosclerosis,
   abeta40_levels_of_patients_with_severe_atherosclerosis
)
```

We displayed the ANOVA results on the bar graph along with the data.

```python
plt.text(
   0.02, 0.95, #coordinates for the text, along with line 217
   "One-Way Anova\n" #imputes the anova values
   f"Aβ40: F = {f_abeta40:.3f}, p = {p_abeta40:.3f}\n"
   f"Aβ42: F = {f_abeta42:.3f}, p = {p_abeta42:.3f}\n"
   f"tTAU: F = {f_ttau:.3f}, p = {p_ttau:.3f}\n"
   f"pTAU: F = {f_ptau:.3f}, p = {p_ptau:.3f}",
   transform=plt.gca().transAxes,
   verticalalignment ="top" #coordinates correspond to top edge of text
)
plt.show()
```
