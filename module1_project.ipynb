## Disease Background:

* Prevalence & incidence:

It’s estimated that approximately 7.4 million Americans have Alzheimer’s disease, about 11% of all people aged 65 and older. The prevalence by age group is 5.2% of all people aged 65-74, 13.8% of 75-84 year olds, all the way up to 35.8% of people over the age of 85. It’s unlikely, but not impossible, for Alzheimer’s symptoms to develop under the age of 65 because of the significant impact between again and the disease. Based on the growth rate of the prevalence, it’s also estimated that the number will almost double every 20 years.
https://www.alz.org/getmedia/ef8f48f9-ad36-48ea-87f9-b74034635c1e/alzheimers-facts-and-figures.pdf
* Economic burden:

According to a study published in PubMed, the annual per capita cost of Alzheimer’s disease ranges from $468.28 to $171,283.80, based on severity. The vast majority of this figure comes from the cost of caregiving, because people with more severe Alzheimer’s are often incapable of caring for themselves. The majority of these costs are paid by Medicare/Medicaid, with only a small amount covered by private insurance, and the remaining out of pocket. Caring for a relative with Alzheimer's is a significant economic burden, and can often require around-the-clock care, causing someone to either have to hire an expensive caregiver or sacrifice work hours or even leave their jobs to be able to take care of their loved ones.
https://pubmed.ncbi.nlm.nih.gov/37972428/
https://www.alz.org/getmedia/ef8f48f9-ad36-48ea-87f9-b74034635c1e/alzheimers-facts-and-figures.pdf
* Risk factors (genetic, lifestyle):

Because we don’t know exactly how Alzheimer’s starts, we can’t say for certain what causes it, but there are a number of risk factors. Some of the most influential ones are age, genetics, lifestyle, and overall health. Age is the most significant risk factor, with 11% of people 65% affected. Genetics can also impact a person's likelihood of developing Alzheimer's, as variants of many genes have been found that increase or decrease the risk. In addition, exercise, diet, and cognitive engagement can reduce risk of Alzheimer’s by improving overall brain health. 
https://www.alz.org/getmedia/ef8f48f9-ad36-48ea-87f9-b74034635c1e/alzheimers-facts-and-figures.pdf
* Societal determinants:

Lower socioeconomic status can increase risk of Alzheimer’s, mainly because of the overall health impacts. As discussed above, healthier lifestyles can prevent Alzheimer’s, and lower socioeconomic status means less access to exercise, healthy food, medecine, and more. In addition, those with more years of formal education are at lower risk of developing Alzheimer's because they have healthier brains. Another, somewhat lesser known, societal determinant is living environment, specifically exposure to fine particulate matter air pollution. Recent studies have reported higher long-term exposure is correlated to worse cognitive decline and severity of Alzheimer's.
https://www.alz.org/getmedia/ef8f48f9-ad36-48ea-87f9-b74034635c1e/alzheimers-facts-and-figures.pdf
* Symptoms:

In Alzheimer’s disease, neurons gradually break down because of the build up of plaque. This causes cognitive impairments such as memory problems, emotional dysfunction, communication problems, and others. In early stages, Alzheimer's generally presents with minor cognitive decline and subtle changes in mood or behavior. By later stages of the disease symptoms can even include difficulty walking, speaking, and swallowing because of the damage to the cerebral cortex.
https://www.alz.org/getmedia/ef8f48f9-ad36-48ea-87f9-b74034635c1e/alzheimers-facts-and-figures.pdf

## Data-Set:
The data set we will analyze is a collection of information on almost 100 patients from two studies. It contains information about the patients such as demographics and societal determinants, several different measures of cognitive status, information about their brains such as pH and weight, concentrations of Amyloid-Beta and Tau proteins, other health conditions, and more. Some of the data was self-reported by the patients, but much of it was collected after death by examining the patient brains. The Amyloid-Beta and Tau concentrations are measured in pg/ug.
